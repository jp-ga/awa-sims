```python
from xopt.vocs import VOCS
from xopt.evaluator import Evaluator
from xopt.generators.bayesian import ExpectedImprovementGenerator
from xopt import Xopt
from opal import evaluate_opal_awa
```

```python
vocs = VOCS(
    variables = {
        #'dq4_k1': [0.336668-0.2, 0.336668+0.2],
        #'dq5_k1': [-0.470359-0.2, -0.470359+0.2],
        #'dq6_k1': [0.436327-0.2, 0.436327+0.2],
        #'dq7_k1': [-0.429156-0.2, -0.429156+0.2],
        'rq_k1': [-2.1-1.0, -2.1+1.0],
    },
    objectives = {
        'std_x': "MINIMIZE",
    },
    #constraints = { 'particle_loss': ['LESS_THAN', 0.01] },
    constants = {
        'template_name': 'cathode_to_rquad_midpoint.in',
        'num_processors': 4,
        'sol_focusing_i': 550.0,
        'sol_matching_i': 200.0,
        'gun_phase': 50.0,
        'linac_phase': -10.0,
        'linac_1_sol_i': 0.0,
        'linac_2_sol_i': 0.0,
        'linac_3_sol_i': 0.0,
        'dq4_k1': 0.261428,
        'dq5_k1': -0.670359,
        'dq6_k1': 0.621865,
        'dq7_k1': -0.309544,
        #'rq1_k1': 0.0,
        'charge': 1.0e-9,
        'laser_radius': 0.002,
        'csr_on': 0,
        'gun_field_maps': 2,
        'linac_field_maps': 2,
        'autophase': 2,
        'ps_dump_freq': 100000000,
        'stat_dump_freq': 100000000,
        'repart_freq': 100000000,
        'grid_points': 8
    }
)
```

```python
vocs
```

```python
dump_file = 'opt_dump.yml'
evaluator = Evaluator(function=evaluate_opal_awa)
generator = ExpectedImprovementGenerator(vocs=vocs)
X = Xopt(dump_file=dump_file, evaluator=evaluator, generator=generator, vocs=vocs)
```

```python
X.random_evaluate(1)
```

```python
for i in range(50):
    print(f'Iteration {i}')
    X.step()
```

```python
best_id, best_beamsize, best_ks = X.vocs.select_best(X.data)
print(best_id, best_beamsize, best_ks)
```
