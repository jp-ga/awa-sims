```python
from xopt.vocs import VOCS
from xopt.evaluator import Evaluator
from xopt.generators.bayesian import ExpectedImprovementGenerator
from xopt import Xopt
from opal import evaluate_opal_awa
```

```python
#focusing_sol_i = 550.0
matching_sol_i = 320.0 
delta_i = 30.0 
```

```python
vocs = VOCS(
    variables = {
        #'focusing_sol_i': [focusing_sol_i - delta_i, focusing_sol_i + delta_i],
        #'matching_sol_i': [0.0, 440.0],
        'linac1_sol_i': [0.0, 500.0],
    },
    objectives = {
        #'beam_size': 'MINIMIZE',
        #'beam_size_3d': 'MINIMIZE'
        #'std_z': "MINIMIZE"
        'particle_loss': "MINIMIZE"
    },
    #constraints = { 'particle_loss': ['LESS_THAN', 0.01] },
    constants = {
        'template_name': 'cathode_to_EYG5.in',
        'num_processors': 4,
        'focusing_sol_i': 550.0,
        'matching_sol_i': 210.0,
        'gun_phase': 50.0, 
        'linac_phase': -10.0,
        #'linac1_sol_i': 200.0,
        'linac2_sol_i': 0.0,
        'linac3_sol_i': 0.0,
        'dq4_k1': 0.0,
        'dq5_k1': 0.0,
        'dq6_k1': 0.0,
        'dq7_k1': 0.0,
        'rq1_k1': 0.0,
        'rq2_k1': 0.0,
        'n_particles': 100000,
        'charge': 10.0e-9,
        'laser_radius': 0.005,
        'csr_on': 1,
        'gun_field_maps': 3,
        'linac_field_maps': 3,
        'autophase': 2,
        'ps_dump_freq': 10000000,
        'stat_dump_freq': 100,
        'repart_freq': 10,
        'mx': 8,
        'my': 8,
        'mt': 8,
    }
)
```

```python
vocs
```

```python
dump_file = 'opt_dump.yml'
evaluator = Evaluator(function=evaluate_opal_awa)
generator = ExpectedImprovementGenerator(vocs=vocs)
X = Xopt(dump_file=dump_file, evaluator=evaluator, generator=generator, vocs=vocs)
```

```python
X.random_evaluate(1)
```

```python
for i in range(50):
    print(f'Iteration {i}')
    X.step()
```

```python
best_id, best_beamsize, best_ks = X.vocs.select_best(X.data)
print(best_id, best_beamsize, best_ks)
```
