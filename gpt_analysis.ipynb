```python
import numpy as np
import matplotlib.pyplot as plt
import os
import easygdf
```

```python
run_directory = '../data/gpt/saved_runs/test'
```

```python
cathode_screen = easygdf.load_screens_touts(
    os.path.join(run_directory, 'out.gdf')
)['screens'][0]
```

```python
fig, ax = plt.subplots()
ax.hist2d(cathode_screen['x']*1e3, cathode_screen['y']*1e3, bins=100)
ax.set_xlabel('$x$ (mm)')
ax.set_ylabel('$y$ (mm)')
ax.set_aspect('equal')
plt.show()
```
