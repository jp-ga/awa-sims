```python
import numpy as np
import matplotlib.pyplot as plt
import os
import easygdf
```

```python
run_directory = '../data/gpt/saved_runs/1nc'
```

```python
screens = easygdf.load_screens_touts(
    os.path.join(run_directory, 'out.gdf')
)['screens']
screens
```

```python
cathode_screen = screens[1]
```

```python
fig, ax = plt.subplots()
ax.hist2d(cathode_screen['x']*1e3, cathode_screen['y']*1e3, bins=100)
ax.set_xlabel('$x$ (mm)')
ax.set_ylabel('$y$ (mm)')
ax.set_aspect('equal')
plt.show()
```

```python
yag4 = screens[0]
```

```python
plt.hist2d(
    (yag4['t'] - yag4['t'].mean())*1e12, 
    yag4['Bz']*yag4['G'],
    bins=200,
    cmin=1
)
plt.xlabel('$t$ (ps)')
plt.ylabel('$\gamma \\beta_z$')
plt.show()
```

```python
plt.hist2d(
    yag4['x']*1e3, 
    yag4['Bx']/yag4['Bz']*1e3,
    bins=200,
    cmin=1
)
plt.xlabel('$x$ (mm)')
plt.ylabel('$\gamma \\beta_x \\times 10^3$')
plt.show()
```
